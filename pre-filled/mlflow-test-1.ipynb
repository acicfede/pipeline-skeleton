```python
import mlflow
import random

mlflow.set_tracking_uri("https://mlflow-monitoring-tools.apps.privateailab.acic.accenture")

experiment_name = "ds_dashboard_demo"
if mlflow.get_experiment_by_name(experiment_name) is None:
    mlflow.create_experiment(experiment_name)
mlflow.set_experiment(experiment_name)

with mlflow.start_run():
    for step in range(5):
        # Metriche di classificazione simulate
        mlflow.log_metric("accuracy", random.random(), step=step)
        mlflow.log_metric("precision", random.random(), step=step)
        mlflow.log_metric("recall", random.random(), step=step)
        mlflow.log_metric("f1_score", random.random(), step=step)
        mlflow.log_metric("roc_auc", random.random(), step=step)

        # Metriche generiche
        mlflow.log_metric("training_loss", random.random(), step=step)
        mlflow.log_metric("validation_loss", random.random(), step=step)
        mlflow.log_metric("learning_rate", 0.001, step=step)

        # Metriche custom / random
        mlflow.log_metric("random_metric", random.random(), step=step)

print("Run loggato con tutte le metriche!")
```
