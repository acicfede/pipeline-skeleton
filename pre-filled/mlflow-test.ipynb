```python
# dataset di esempio
X = np.array([[1], [2], [3], [4]])
y = np.array([2, 4, 6, 8])

with mlflow.start_run(run_name="linear-regression-notebook"):
    model = LinearRegression().fit(X, y)
    
    # parametri
    mlflow.log_param("fit_intercept", True)

    # metriche
    score = model.score(X, y)
    mlflow.log_metric("score", score)

    # salva modello come artifact
    mlflow.sklearn.log_model(model, "model")

score
```
